```python
import re
import os
import pandas as pd
```

```python
site_list = ['FR-Cap', 'UK-MCR']
scenario_list = ['cntl', 'traffic']
path = 'timing/'
all_rows = []   # store all timing rows here

for site in site_list:
    for scenario in scenario_list:
        filename = f"{site}_{scenario}"
        matching_files = [f for f in os.listdir(path) if filename in f]

        for file in matching_files:
            filepath = os.path.join(path, file)

            timing_list = []
            with open(filepath, "r") as f:
                for i, line in enumerate(f, start=1):
                    if i in [36, 37, 38, 46, 47]:
                        # extract the *first* number in the line
                        nums = re.findall(r"[-+]?\d*\.\d+|\d+", line)
                        if nums:
                            timing_list.append(float(nums[0]))

            # Ensure we got all 5 expected values
            if len(timing_list) == 5:
                all_rows.append(
                    [site, scenario] + timing_list
                )

# Create final DataFrame
df_timing = pd.DataFrame(
    all_rows,
    columns=[
        'Site', 'Scenario',
        'Init Time', 'Run Time', 'Final Time',
        'ATM Run Time', 'LND Run Time'
    ]
)

df_timing['CPL Run Time'] = df_timing['Run Time'] - df_timing['ATM Run Time'] - df_timing['LND Run Time']
output_file = 'data_for_figure/model_timing_summary.csv'
df_timing.to_csv(output_file, index=False)
df_timing
```
